```python
# DAY 2 - USER INPUT SYSTEM

print("================================")
print("       AI SCHEDULE PLANNER")
print("================================")

name = input("Enter your name: ")
subject = input("Enter subject name: ")
priority = input("Enter priority (High/Medium/Low): ")
difficulty = input("Enter difficulty (Easy/Medium/Hard): ")
deadline = input("Enter deadline (DD-MM-YYYY): ")
study_time = float(input("Enter available study time (hours): "))

print("\n----- YOUR INPUT -----")
print("Name:", name)
print("Subject:", subject)
print("Priority:", priority)
print("Difficulty:", difficulty)
print("Deadline:", deadline)
print("Study Time:", study_time, "hours")
```

```python
# Store the subject details in a dictionary

subject_data = {
    "name": name,
    "subject": subject,
    "priority": priority,
    "difficulty": difficulty,
    "deadline": deadline,
    "study_time": study_time
}

print("Subject data stored successfully!")
print(subject_data)
```

```python
# Display subject information

print("----- SUBJECT INFORMATION -----")
print("Name:", subject_data["name"])
print("Subject:", subject_data["subject"])
print("Priority:", subject_data["priority"])
print("Difficulty:", subject_data["difficulty"])
print("Deadline:", subject_data["deadline"])
print("Study Time:", subject_data["study_time"], "hours")
```

```python
# Store subject data in a list

subjects = []

subjects.append(subject_data)

print("----- ALL SUBJECTS -----")

for i, sub in enumerate(subjects, start=1):
    print("Subject", i)
    print("Name:", sub["name"])
    print("Subject:", sub["subject"])
    print("Priority:", sub["priority"])
    print("Difficulty:", sub["difficulty"])
    print("Deadline:", sub["deadline"])
    print("Study Time:", sub["study_time"], "hours")
    print("------------------------")
```

```python
# Day 3 - Add Multiple Subjects

subjects = []

number = int(input("How many subjects do you want to add? "))

for i in range(number):
    print("\nEnter details for Subject", i + 1)

    subject = input("Subject name: ")
    priority = input("Priority (High/Medium/Low): ")
    study_time = float(input("Study time in hours: "))

    subject_data = {
        "subject": subject,
        "priority": priority,
        "study_time": study_time
    }

    subjects.append(subject_data)

print("\n===== ALL SUBJECTS =====")

for i, sub in enumerate(subjects, start=1):
    print("\nSubject", i)
    print("Name:", sub["subject"])
    print("Priority:", sub["priority"])
    print("Study Time:", sub["study_time"], "hours")
```

```python
# Day 3 - Calculate Total Study Time

total_time = 0

for sub in subjects:
    total_time += sub["study_time"]

print("===== STUDY TIME SUMMARY =====")
print("Total Subjects:", len(subjects))
print("Total Study Time:", total_time, "hours")
```

```python
# Day 3 - Count High Priority Subjects

high_priority_count = 0

for sub in subjects:
    if sub["priority"].lower() == "high":
        high_priority_count += 1

print("===== PRIORITY SUMMARY =====")
print("High Priority Subjects:", high_priority_count)
```

```python
# Day 3 - Count Hard Subjects

hard_count = 0

for sub in subjects:
    if sub.get("difficulty", "").lower() == "hard":
        hard_count += 1

print("===== DIFFICULTY SUMMARY =====")
print("Hard Subjects:", hard_count)
```

```python
# Day 3 - Count Medium Priority Subjects

medium_count = 0

for sub in subjects:
    if sub.get("priority", "").lower() == "medium":
        medium_count += 1

print("===== MEDIUM PRIORITY SUMMARY =====")
print("Medium Priority Subjects:", medium_count)
```

```python
# Day 3 - Count Easy Subjects

easy_count = 0

for sub in subjects:
    if sub.get("difficulty", "").lower() == "easy":
        easy_count += 1

print("===== DIFFICULTY SUMMARY =====")
print("Easy Subjects:", easy_count)
```

```python
# Day 3 - Count Medium Difficulty Subjects

medium_difficulty_count = 0

for sub in subjects:
    if sub.get("difficulty", "").lower() == "medium":
        medium_difficulty_count += 1

print("===== DIFFICULTY SUMMARY =====")
print("Medium Difficulty Subjects:", medium_difficulty_count)
```

```python
# Day 3 - Count Hard Difficulty Subjects

hard_difficulty_count = 0

for sub in subjects:
    if sub.get("difficulty", "").lower() == "hard":
        hard_difficulty_count += 1

print("===== DIFFICULTY SUMMARY =====")
print("Hard Difficulty Subjects:", hard_difficulty_count)
```

```python
# Day 3 - Final Study Planner Summary

print("===== AI SCHEDULE PLANNER =====")
print("===== FINAL SUMMARY =====")

print("Total Subjects:", len(subjects))
print("Total Study Time:", total_time, "hours")
print("High Priority Subjects:", high_priority_count)
print("Medium Priority Subjects:", medium_count)
print("Easy Difficulty Subjects:", easy_count)
print("Medium Difficulty Subjects:", medium_difficulty_count)
print("Hard Difficulty Subjects:", hard_difficulty_count)
```
