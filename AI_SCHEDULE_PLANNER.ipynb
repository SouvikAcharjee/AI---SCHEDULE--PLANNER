```python
# DAY 5 - DEADLINE-BASED STUDY PLANNER

from datetime import datetime

subjects = [
    {
        "subject": "Python",
        "priority": "High",
        "study_time": 2,
        "deadline": "15-10-2026"
    },
    {
        "subject": "Mathematics",
        "priority": "Medium",
        "study_time": 1.5,
        "deadline": "12-10-2026"
    },
    {
        "subject": "Java",
        "priority": "Low",
        "study_time": 1,
        "deadline": "20-10-2026"
    }
]

subjects.sort(
    key=lambda sub: datetime.strptime(
        sub["deadline"], "%d-%m-%Y"
    )
)

print("===== DAY 5: DEADLINE-WISE STUDY PLAN =====")

for i, sub in enumerate(subjects, start=1):
    print(
        i, ".", sub["subject"],
        "| Deadline:", sub["deadline"],
        "| Priority:", sub["priority"],
        "| Study Time:", sub["study_time"], "hours"
    )
```

```python
# DAY 5 - DAYS REMAINING

from datetime import datetime

today = datetime.today().date()

print("===== DEADLINE COUNTDOWN =====")

for sub in subjects:
    deadline = datetime.strptime(
        sub["deadline"], "%d-%m-%Y"
    ).date()

    days_left = (deadline - today).days

    if days_left < 0:
        status = "Deadline Passed"
    elif days_left == 0:
        status = "Deadline Today!"
    else:
        status = str(days_left) + " days remaining"

    print(sub["subject"], "-", status)
```

```python
# DAY 5 - URGENT SUBJECTS

from datetime import datetime

today = datetime.today().date()

print("===== URGENT SUBJECTS =====")

urgent_found = False

for sub in subjects:
    deadline = datetime.strptime(
        sub["deadline"], "%d-%m-%Y"
    ).date()

    days_left = (deadline - today).days

    if 0 <= days_left <= 3:
        print(
            "URGENT:", sub["subject"],
            "| Deadline in", days_left, "day(s)"
        )
        urgent_found = True

if not urgent_found:
    print("No urgent deadlines!")

print("\nUrgent subject check completed!")
```

```python
# DAY 5 - FINAL STUDY PLAN

from datetime import datetime

today = datetime.today().date()
total_time = 0

print("===== AI SCHEDULE PLANNER =====")
print("===== DAY 5 FINAL REPORT =====")

for sub in subjects:
    deadline = datetime.strptime(
        sub["deadline"], "%d-%m-%Y"
    ).date()

    days_left = (deadline - today).days
    total_time += sub["study_time"]

    if days_left < 0:
        status = "OVERDUE"
    elif days_left <= 3:
        status = "URGENT"
    else:
        status = "ON TRACK"

    print(
        sub["subject"], "|",
        status, "|",
        sub["study_time"], "hours"
    )

print("------------------------------")
print("Total Subjects:", len(subjects))
print("Total Study Time:", total_time, "hours")
print("Day 5 completed successfully!")
```

```python
# DAY 6 - STUDY TIME RECOMMENDATION

def recommend_study_time(priority, difficulty):
    if priority.lower() == "high" and difficulty.lower() == "hard":
        return 3
    elif priority.lower() == "high" or difficulty.lower() == "hard":
        return 2
    elif priority.lower() == "medium" or difficulty.lower() == "medium":
        return 1.5
    else:
        return 1

print("===== STUDY TIME RECOMMENDATION =====")

print("Python (High, Hard):",
      recommend_study_time("High", "Hard"), "hours")

print("Mathematics (Medium, Medium):",
      recommend_study_time("Medium", "Medium"), "hours")

print("English (Low, Easy):",
      recommend_study_time("Low", "Easy"), "hours")
```

```python
# DAY 6 - STUDY TIME COMPARISON

available_time = float(
    input("Enter available study time (hours): ")
)

print("\n===== STUDY TIME COMPARISON =====")
print("Recommended Time:", recommended_time, "hours")
print("Available Time:", available_time, "hours")

if available_time >= recommended_time:
    print("Great! You have enough time to study.")
else:
    shortage = recommended_time - available_time
    print("You need", shortage, "more hours to complete the study plan.")

print("Study time analysis completed!")
```

```python
# DAY 7 - STUDY PROGRESS TRACKER

subject = input("Enter subject name: ")
planned_hours = float(input("Enter planned study hours: "))
completed_hours = float(input("Enter completed study hours: "))

print("\n===== STUDY PROGRESS REPORT =====")
print("Subject:", subject)
print("Planned Study Time:", planned_hours, "hours")
print("Completed Study Time:", completed_hours, "hours")

remaining_hours = planned_hours - completed_hours

if planned_hours > 0:
    progress = (completed_hours / planned_hours) * 100
else:
    progress = 0

print("Remaining Study Time:", max(0, remaining_hours), "hours")
print("Progress:", round(progress, 2), "%")

if completed_hours >= planned_hours:
    print("Status: Study goal completed!")
else:
    print("Status: Keep going!")

print("Day 7 - Progress tracking completed!")
```

```python
# DAY 7 - PROGRESS STATUS CHECKER

print("===== PROGRESS STATUS CHECKER =====")

if progress >= 100:
    print("Excellent! Your study goal is complete!")
elif progress >= 75:
    print("Great progress! Almost there!")
elif progress >= 50:
    print("Good progress! Keep going!")
elif progress > 0:
    print("You have started. Stay consistent!")
else:
    print("Start studying now!")

print("Current Progress:", round(progress, 2), "%")
print("Keep learning every day!")
```

```python
# DAY 7 - VISUAL PROGRESS BAR

print("===== STUDY PROGRESS BAR =====")

bar_length = 20
filled = int(bar_length * min(progress, 100) / 100)
bar = "█" * filled + "░" * (bar_length - filled)

print("[" + bar + "]", round(progress, 2), "%")

print("Completed:", min(completed_hours, planned_hours), "hours")
print("Planned:", planned_hours, "hours")
```

```python
# DAY 7 - FINAL PROGRESS REPORT

print("================================")
print("   AI SCHEDULE PLANNER")
print("   DAY 7 - FINAL REPORT")
print("================================")

print("Subject:", subject)
print("Planned Hours:", planned_hours)
print("Completed Hours:", completed_hours)
print("Remaining Hours:", max(0, planned_hours - completed_hours))
print("Progress:", round(progress, 2), "%")

if completed_hours >= planned_hours:
    print("Final Status: Goal Completed!")
else:
    print("Final Status: Keep Practicing!")

print("Day 7 completed successfully!")
```

```python
# DAY 8 - DAILY STUDY SCHEDULE GENERATOR

study_schedule = [
    {"subject": "Python", "hours": 2},
    {"subject": "Mathematics", "hours": 1.5},
    {"subject": "English", "hours": 1}
]

start_hour = 9

print("===== MY DAILY STUDY TIMETABLE =====")

for item in study_schedule:
    end_hour = start_hour + item["hours"]

    print(
        item["subject"], ":",
        start_hour, "to", end_hour,
        "| Duration:", item["hours"], "hours"
    )

    start_hour = end_hour

print("\nDaily timetable generated successfully!")
```

```python
# DAY 8 - TIMETABLE WITH BREAKS

study_schedule = [
    {"subject": "Python", "hours": 2},
    {"subject": "Mathematics", "hours": 1.5},
    {"subject": "English", "hours": 1}
]

start_hour = 9
break_minutes = 15

print("===== DAILY STUDY TIMETABLE =====")

for item in study_schedule:
    duration = item["hours"]
    end_hour = start_hour + duration

    print(
        f"{item['subject']}: {start_hour:.2f} to {end_hour:.2f}"
    )

    start_hour = end_hour + break_minutes / 60
    print("Break:", break_minutes, "minutes")

print("\nTimetable with breaks generated!")
```

```python
# DAY 8 - TIMETABLE WITH CLOCK FORMAT

from datetime import datetime, timedelta

study_schedule = [
    {"subject": "Python", "hours": 2},
    {"subject": "Mathematics", "hours": 1.5},
    {"subject": "English", "hours": 1}
]

current_time = datetime.strptime("09:00 AM", "%I:%M %p")
break_minutes = 15

print("===== DAILY STUDY TIMETABLE =====")

for item in study_schedule:
    start_time = current_time
    end_time = start_time + timedelta(hours=item["hours"])

    print(
        item["subject"], ":",
        start_time.strftime("%I:%M %p"),
        "to",
        end_time.strftime("%I:%M %p")
    )

    current_time = end_time + timedelta(minutes=break_minutes)
    print("Break:", break_minutes, "minutes")

print("\nTimetable generated successfully!")
```

```python
# DAY 8 - DAILY SCHEDULE SUMMARY

total_study_hours = sum(
    item["hours"] for item in study_schedule
)

total_break_minutes = break_minutes * len(study_schedule)

print("===== DAY 8 SCHEDULE SUMMARY =====")
print("Total Subjects:", len(study_schedule))
print("Total Study Time:", total_study_hours, "hours")
print("Total Break Time:", total_break_minutes, "minutes")
print("Total Planned Time:",
      total_study_hours + total_break_minutes / 60, "hours")

print("\nDay 8 completed successfully!")
```

```python
# DAY 10 - PRIORITY-BASED TIMETABLE

from datetime import datetime, timedelta

study_schedule = [
    {"subject": "Python", "priority": "High", "hours": 2},
    {"subject": "Mathematics", "priority": "Medium", "hours": 1.5},
    {"subject": "English", "priority": "Low", "hours": 1}
]

priority_order = {
    "high": 1,
    "medium": 2,
    "low": 3
}

study_schedule.sort(
    key=lambda item: priority_order.get(
        item["priority"].lower(), 4
    )
)

current_time = datetime.strptime("09:00 AM", "%I:%M %p")
break_minutes = 15

print("===== DAY 10: PRIORITY TIMETABLE =====")

for item in study_schedule:
    start_time = current_time
    end_time = start_time + timedelta(hours=item["hours"])

    print(
        item["subject"], "|",
        item["priority"], "Priority |",
        start_time.strftime("%I:%M %p"),
        "to", end_time.strftime("%I:%M %p")
    )

    current_time = end_time + timedelta(minutes=break_minutes)

print("\nPriority-based timetable generated!")
```

```python
# DAY 10 - PRIORITY SUMMARY

total_study_hours = sum(
    item["hours"] for item in study_schedule
)

print("===== PRIORITY SUMMARY =====")
print("Total Subjects:", len(study_schedule))
print("Total Study Time:", total_study_hours, "hours")

for priority in ["High", "Medium", "Low"]:
    count = sum(
        1 for item in study_schedule
        if item["priority"].lower() == priority.lower()
    )
    print(priority, "Priority Subjects:", count)

print("\nPriority summary completed!")
```

```python
# DAY 10 - SCHEDULE VALIDATION

valid_priorities = ["high", "medium", "low"]

print("===== SCHEDULE VALIDATION =====")

for item in study_schedule:
    if item["hours"] <= 0:
        print(item["subject"], "- Invalid study hours!")
    elif item["priority"].lower() not in valid_priorities:
        print(item["subject"], "- Invalid priority!")
    else:
        print(item["subject"], "- Valid schedule")

print("\nSchedule validation completed!")
```

```python
# DAY 10 - FINAL SCHEDULE REPORT

print("================================")
print("     AI SCHEDULE PLANNER")
print("     DAY 10 FINAL REPORT")
print("================================")

print("Total Subjects:", len(study_schedule))
print("Total Study Time:", total_study_hours, "hours")

print("\n===== FINAL TIMETABLE =====")

for i, item in enumerate(study_schedule, start=1):
    print(
        i, ".", item["subject"],
        "|", item["priority"], "Priority",
        "|", item["hours"], "hours"
    )

print("\nDay 10 completed successfully!")
```

```python
# DAY 11 - STUDY TASK MANAGER

tasks = []

number = int(input("How many study tasks do you want to add? "))

for i in range(number):
    print("\nTask", i + 1)

    task_name = input("Enter task name: ")
    subject = input("Enter subject: ")
    duration = float(input("Enter duration in hours: "))

    task = {
        "task_name": task_name,
        "subject": subject,
        "duration": duration,
        "completed": False
    }

    tasks.append(task)

print("\n===== YOUR STUDY TASKS =====")

for i, task in enumerate(tasks, start=1):
    print(
        i, ".", task["task_name"],
        "| Subject:", task["subject"],
        "| Duration:", task["duration"], "hours",
        "| Status: Pending"
    )

print("\nStudy tasks added successfully!")
```

```python
# DAY 11 - MARK TASKS AS COMPLETED

print("===== YOUR STUDY TASKS =====")

for i, task in enumerate(tasks, start=1):
    print(i, ".", task["task_name"], "-",
          "Completed" if task["completed"] else "Pending")

task_number = int(input("\nEnter task number to complete: "))

if 1 <= task_number <= len(tasks):
    tasks[task_number - 1]["completed"] = True
    print("\nTask marked as completed!")
else:
    print("Invalid task number!")

print("\n===== UPDATED TASK LIST =====")

for i, task in enumerate(tasks, start=1):
    status = "Completed" if task["completed"] else "Pending"
    print(i, ".", task["task_name"], "| Status:", status)
```

```python
# DAY 12 - SMART STUDY RECOMMENDATIONS

def recommend_study_hours(priority, difficulty):
    if priority == "High" and difficulty == "Hard":
        return 3
    elif priority == "High" or difficulty == "Hard":
        return 2
    elif priority == "Medium" or difficulty == "Medium":
        return 1.5
    else:
        return 1

print("===== SMART STUDY RECOMMENDER =====")

subject = input("Enter subject name: ")
priority = input("Enter priority (High/Medium/Low): ").capitalize()
difficulty = input("Enter difficulty (Hard/Medium/Easy): ").capitalize()

recommended_hours = recommend_study_hours(priority, difficulty)

print("\n===== STUDY RECOMMENDATION =====")
print("Subject:", subject)
print("Priority:", priority)
print("Difficulty:", difficulty)
print("Recommended Study Time:", recommended_hours, "hours")
print("Recommendation generated successfully!")
```

```python
# DAY 12 - FINAL SMART STUDY REPORT

print("===== AI SCHEDULE PLANNER =====")
print("===== DAY 12 FINAL REPORT =====")

print("Subject:", subject)
print("Priority:", priority)
print("Difficulty:", difficulty)
print("Recommended Hours:", recommended_hours)
print("Available Hours:", available_hours)

if available_hours >= recommended_hours:
    print("Final Status: Study plan fits your available time.")
else:
    print("Final Status: Adjust your study plan to fit your time.")

print("\nDay 12 completed successfully!")
```

```python
# DAY 13 - STUDY PROGRESS TRACKER

progress_data = []

number = int(input("How many subjects do you want to track? "))

for i in range(number):
    print("\nSubject", i + 1)

    subject = input("Enter subject name: ")
    planned = float(input("Enter planned study hours: "))
    completed = float(input("Enter completed study hours: "))

    if planned <= 0 or completed < 0 or completed > planned:
        print("Invalid hours! Please enter valid values.")
        continue

    remaining = planned - completed
    percentage = (completed / planned) * 100

    progress_data.append({
        "subject": subject,
        "planned": planned,
        "completed": completed,
        "remaining": remaining,
        "percentage": percentage
    })

print("\n===== STUDY PROGRESS DATA =====")

for item in progress_data:
    print("Subject:", item["subject"])
    print("Planned Hours:", item["planned"])
    print("Completed Hours:", item["completed"])
    print("Remaining Hours:", item["remaining"])
    print("Progress:", round(item["percentage"], 2), "%")

print("\nStudy progress data collected successfully!")
```

```python
# DAY 13 - OVERALL PROGRESS SUMMARY

total_planned = sum(item["planned"] for item in progress_data)
total_completed = sum(item["completed"] for item in progress_data)
total_remaining = total_planned - total_completed

overall_progress = (total_completed / total_planned) * 100

print("===== OVERALL STUDY SUMMARY =====")
print("Total Planned Hours:", total_planned)
print("Total Completed Hours:", total_completed)
print("Total Remaining Hours:", total_remaining)
print("Overall Progress:", round(overall_progress, 2), "%")
```

```python
# DAY 13 - FINAL STUDY PROGRESS REPORT

print("===== AI SCHEDULE PLANNER =====")
print("===== DAY 13 FINAL REPORT =====")

for item in progress_data:
    print("\nSubject:", item["subject"])
    print("Planned Hours:", item["planned"])
    print("Completed Hours:", item["completed"])
    print("Remaining Hours:", item["remaining"])
    print("Progress:", round(item["percentage"], 2), "%")

print("\n===== OVERALL PERFORMANCE =====")
print("Total Planned:", total_planned, "hours")
print("Total Completed:", total_completed, "hours")
print("Overall Progress:", round(overall_progress, 2), "%")

print("\nDay 13 completed successfully!")
```

```python
# DAY 14 - SUBJECT PERFORMANCE ANALYZER

print("===== SUBJECT PERFORMANCE ANALYZER =====")

for item in progress_data:
    subject = item["subject"]
    percentage = item["percentage"]

    print("\nSubject:", subject)
    print("Progress:", round(percentage, 2), "%")

    if percentage >= 80:
        print("Performance: Excellent")
    elif percentage >= 50:
        print("Performance: Good")
    else:
        print("Performance: Needs More Focus")

print("\nSubject analysis completed!")
```

```python
# DAY 14 - WEAKEST SUBJECT IDENTIFICATION

if len(progress_data) > 0:

    weakest_subject = min(
        progress_data,
        key=lambda x: x["percentage"]
    )

    print("===== WEAKEST SUBJECT =====")
    print("Subject:", weakest_subject["subject"])
    print("Progress:", weakest_subject["percentage"], "%")
    print("Remaining Hours:", weakest_subject["remaining"])
    print("Recommendation: Give extra study time.")

else:
    print("No subject data available.")
```

```python
# DAY 14 - STUDY RECOMMENDATION

print("===== SMART STUDY RECOMMENDATION =====")

for item in progress_data:

    if item["percentage"] < 30:
        recommendation = "Study for 3 extra hours"
    elif item["percentage"] < 60:
        recommendation = "Study for 2 extra hours"
    elif item["percentage"] < 80:
        recommendation = "Study for 1 extra hour"
    else:
        recommendation = "Maintain your performance"

    print(item["subject"], ":", recommendation)
```

```python
# DAY 14 - OVERALL PERFORMANCE

average_progress = sum(
    item["percentage"] for item in progress_data
) / len(progress_data) if progress_data else 0

print("===== OVERALL PERFORMANCE =====")
print("Average Progress:", round(average_progress, 2), "%")

if average_progress >= 80:
    print("Overall Status: Excellent")
elif average_progress >= 50:
    print("Overall Status: Good")
else:
    print("Overall Status: Needs Improvement")
```

```python
# DAY 14 - FINAL ANALYTICS REPORT

print("\n===== AI SCHEDULE PLANNER =====")
print("===== DAY 14 ANALYTICS REPORT =====")

for item in progress_data:
    print(
        item["subject"],
        "-",
        round(item["percentage"], 2),
        "% Complete"
    )

print("\nTotal Subjects:", len(progress_data))
print("Average Progress:", round(average_progress, 2), "%")
print("Day 14 completed successfully!")
```

```python
# DAY 15 - AUTOMATIC STUDY TIMETABLE

daily_schedule = []

for item in progress_data:
    if item["remaining"] > 0:
        daily_schedule.append({
            "subject": item["subject"],
            "hours": item["remaining"]
        })

print("===== STUDY SCHEDULE DATA =====")

for item in daily_schedule:
    print(item["subject"], "-", item["hours"], "hours")

print("\nSchedule data prepared successfully!")
```

```python
# DAY 15 - SET STUDY STARTING TIME

from datetime import datetime, timedelta

start_time = datetime.strptime("09:00 AM", "%I:%M %p")

print("Study Starting Time:", start_time.strftime("%I:%M %p"))
```

```python
# DAY 15 - GENERATE AUTOMATIC TIMETABLE

generated_timetable = []
current_time = start_time

for item in daily_schedule:

    study_hours = item["hours"]
    end_time = current_time + timedelta(hours=study_hours)

    generated_timetable.append({
        "subject": item["subject"],
        "start": current_time.strftime("%I:%M %p"),
        "end": end_time.strftime("%I:%M %p"),
        "hours": study_hours
    })

    current_time = end_time + timedelta(minutes=15)

print("Automatic timetable generated!")
```

```python
# DAY 15 - DISPLAY TIMETABLE

print("===== YOUR DAILY STUDY TIMETABLE =====")

for item in generated_timetable:
    print("\nSubject:", item["subject"])
    print("Start Time:", item["start"])
    print("End Time:", item["end"])
    print("Duration:", item["hours"], "hours")
    print("-----------------------------")
```

```python
# DAY 15 - TIMETABLE SUMMARY

total_study_hours = sum(
    item["hours"] for item in generated_timetable
)

total_breaks = max(0, len(generated_timetable) - 1) * 15

print("===== DAY 15 FINAL SUMMARY =====")
print("Total Subjects:", len(generated_timetable))
print("Total Study Hours:", total_study_hours)
print("Total Break Time:", total_breaks, "minutes")
print("Timetable Status: Generated Successfully")

print("\nDay 15 completed successfully!")
```

```python
# DAY 16 - SUBJECT PRIORITY MANAGEMENT

priority_subjects = []

number = int(input("How many subjects do you want to add? "))

for i in range(number):
    print("\nSubject", i + 1)

    subject = input("Enter subject name: ")
    priority = input("Enter priority (High/Medium/Low): ").capitalize()

    if priority not in ["High", "Medium", "Low"]:
        print("Invalid priority! Skipping subject.")
        continue

    priority_subjects.append({
        "subject": subject,
        "priority": priority
    })

print("\nPriority subjects added successfully!")
```

```python
# DAY 16 - PRIORITY SCORING

priority_score = {
    "High": 3,
    "Medium": 2,
    "Low": 1
}

for item in priority_subjects:
    item["score"] = priority_score[item["priority"]]

print("Priority scores assigned successfully!")
```

```python
# DAY 16 - SORT SUBJECTS

priority_subjects.sort(
    key=lambda item: item["score"],
    reverse=True
)

print("===== PRIORITY WISE SUBJECTS =====")

for item in priority_subjects:
    print(item["subject"], "-", item["priority"])
```

```python
# DAY 16 - PRIORITY RECOMMENDATIONS

print("===== STUDY RECOMMENDATIONS =====")

for item in priority_subjects:
    if item["priority"] == "High":
        print(item["subject"], ": Study first - High Priority")
    elif item["priority"] == "Medium":
        print(item["subject"], ": Study after high priority subjects")
    else:
        print(item["subject"], ": Study when time is available")
```

```python
# DAY 16 - FINAL PRIORITY REPORT

high_count = sum(1 for x in priority_subjects if x["priority"] == "High")
medium_count = sum(1 for x in priority_subjects if x["priority"] == "Medium")
low_count = sum(1 for x in priority_subjects if x["priority"] == "Low")

print("\n===== DAY 16 FINAL REPORT =====")
print("Total Subjects:", len(priority_subjects))
print("High Priority:", high_count)
print("Medium Priority:", medium_count)
print("Low Priority:", low_count)

print("\nDay 16 completed successfully!")
```

```python
# DAY 17 - DAILY STUDY GOALS

daily_goals = []

number = int(input("How many subjects for today's goal? "))

for i in range(number):
    print("\nSubject", i + 1)

    subject = input("Enter subject name: ")
    target = float(input("Enter target study hours: "))

    daily_goals.append({
        "subject": subject,
        "target": target,
        "completed": 0
    })

print("\nDaily goals created successfully!")
```

```python
# DAY 17 - UPDATE STUDY HOURS

for item in daily_goals:
    print("\nSubject:", item["subject"])

    completed = float(input("Enter completed study hours: "))

    if completed < 0:
        print("Invalid hours! Setting completed hours to 0.")
        completed = 0

    item["completed"] = completed

print("\nStudy hours updated successfully!")
```

```python
# DAY 17 - GOAL ACHIEVEMENT

for item in daily_goals:
    target = item["target"]
    completed = item["completed"]

    if target > 0:
        percentage = (completed / target) * 100
    else:
        percentage = 0

    item["percentage"] = percentage

    print("\nSubject:", item["subject"])
    print("Goal Achievement:", round(percentage, 2), "%")
```

```python
# DAY 17 - GOAL STATUS

print("===== DAILY GOAL STATUS =====")

for item in daily_goals:
    if item["completed"] >= item["target"]:
        print(item["subject"], ": Goal Achieved!")
    else:
        remaining = item["target"] - item["completed"]
        print(item["subject"], ": Remaining", remaining, "hours")
```

```python
# DAY 17 - FINAL REPORT

total_target = sum(x["target"] for x in daily_goals)
total_completed = sum(x["completed"] for x in daily_goals)

achievement = (
    total_completed / total_target * 100
    if total_target > 0 else 0
)

print("\n===== DAY 17 FINAL REPORT =====")
print("Total Target Hours:", total_target)
print("Total Completed Hours:", total_completed)
print("Overall Achievement:", round(achievement, 2), "%")

print("\nDay 17 completed successfully!")
```

```python
# DAY 18 - STUDY REMINDER SYSTEM

reminders = []

number = int(input("How many reminders do you want to set? "))

for i in range(number):
    print("\nReminder", i + 1)

    subject = input("Enter subject name: ")
    study_time = input("Enter study time (e.g. 10:00 AM): ")
    message = input("Enter reminder message: ")

    reminders.append({
        "subject": subject,
        "time": study_time,
        "message": message
    })

print("\nReminders created successfully!")
```

```python
# DAY 18 - DISPLAY REMINDERS

print("===== YOUR STUDY REMINDERS =====")

for i, reminder in enumerate(reminders, start=1):
    print("\nReminder:", i)
    print("Subject:", reminder["subject"])
    print("Time:", reminder["time"])
    print("Message:", reminder["message"])
```

```python
# DAY 18 - SEARCH REMINDER

search_subject = input("Enter subject to search: ").lower()

found = False

for reminder in reminders:
    if reminder["subject"].lower() == search_subject:
        print("\nReminder Found!")
        print("Subject:", reminder["subject"])
        print("Time:", reminder["time"])
        print("Message:", reminder["message"])
        found = True

if not found:
    print("No reminder found for this subject.")
```

```python
# DAY 18 - REMINDER STATUS

for reminder in reminders:
    reminder["completed"] = False

subject_name = input("Enter subject completed: ").lower()

found = False

for reminder in reminders:
    if reminder["subject"].lower() == subject_name:
        reminder["completed"] = True
        print("Reminder marked as completed!")
        found = True

if not found:
    print("Subject not found.")
```

```python
# DAY 18 - FINAL REMINDER REPORT

completed_count = sum(
    1 for r in reminders if r["completed"]
)

pending_count = len(reminders) - completed_count

print("===== DAY 18 FINAL REPORT =====")
print("Total Reminders:", len(reminders))
print("Completed:", completed_count)
print("Pending:", pending_count)

print("\nDay 18 completed successfully!")
```

```python
# DAY 19 - BREAK AND REST MANAGEMENT

study_sessions = []

number = int(input("How many study sessions do you want? "))

for i in range(number):
    print("\nSession", i + 1)

    subject = input("Enter subject name: ")
    hours = float(input("Enter study hours: "))

    study_sessions.append({
        "subject": subject,
        "hours": hours
    })

print("\nStudy sessions created successfully!")
```

```python
# DAY 19 - AUTOMATIC BREAK SYSTEM

for session in study_sessions:
    session["break_minutes"] = 15

print("15-minute breaks added between study sessions.")
```

```python
# DAY 19 - BREAK TIMETABLE

from datetime import datetime, timedelta

current_time = datetime.strptime("09:00 AM", "%I:%M %p")

for session in study_sessions:
    end_time = current_time + timedelta(hours=session["hours"])

    print("\nSubject:", session["subject"])
    print("Start:", current_time.strftime("%I:%M %p"))
    print("End:", end_time.strftime("%I:%M %p"))

    current_time = end_time + timedelta(minutes=15)

print("\nStudy and break timetable generated!")
```

```python
# DAY 19 - TIME CALCULATION

total_study = sum(
    session["hours"] for session in study_sessions
)

total_rest = max(0, len(study_sessions) - 1) * 15

print("Total Study Time:", total_study, "hours")
print("Total Break Time:", total_rest, "minutes")
print("Total Rest Time:", round(total_rest / 60, 2), "hours")
```

```python
# DAY 19 - FINAL REST MANAGEMENT REPORT

print("\n===== DAY 19 FINAL REPORT =====")

print("Total Sessions:", len(study_sessions))
print("Total Study Hours:", total_study)
print("Total Break Minutes:", total_rest)

print("\n===== SESSION DETAILS =====")

for session in study_sessions:
    print(session["subject"], "-", session["hours"], "hours")

print("\nDay 19 completed successfully!")
```
